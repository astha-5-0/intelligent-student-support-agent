```python
# Step 2: Import required libraries

import json
import time
import re
from pathlib import Path
from datetime import datetime

print("All basic libraries imported successfully!")
print("Project setup is ready.")
```

```python
# Step 3: Sample student database

student_records = {
    "STU001": {
        "name": "Aarav",
        "fee_status": "Paid",
        "scholarship_status": "Approved"
    },
    "STU002": {
        "name": "Priya",
        "fee_status": "Pending",
        "scholarship_status": "Under Review"
    },
    "STU003": {
        "name": "Rahul",
        "fee_status": "Partial Payment",
        "scholarship_status": "Not Applied"
    }
}

print("Student records created successfully!")
print("Total sample students:", len(student_records))
print(student_records["STU001"])
```

```python
# Step 4: Tool to check student fee status

def check_fee_status(student_id):
    if student_id in student_records:
        return {
            "student_id": student_id,
            "fee_status": student_records[student_id]["fee_status"]
        }
    else:
        return {"error": "Student record not found."}


# Test the tool
print(check_fee_status("STU001"))
print(check_fee_status("STU002"))
```

```python
# Step 5: Tool to check scholarship status

def check_scholarship_status(student_id):
    if student_id in student_records:
        return {
            "student_id": student_id,
            "scholarship_status": student_records[student_id]["scholarship_status"]
        }
    else:
        return {"error": "Student record not found."}


# Test the tool
print(check_scholarship_status("STU001"))
print(check_scholarship_status("STU002"))
print(check_scholarship_status("STU999"))
```

```python
# Step 6: Student policy document

student_policies = [
    {
        "topic": "Fee Payment",
        "content": "Students should check the official fee portal for their current dues. If payment fails, save the transaction reference and contact the accounts office. The next working day payment rule must be confirmed with the official accounts office."
    },
    {
        "topic": "Scholarship",
        "content": "Scholarship applications are subject to eligibility checks and institutional approval. Students should check their application status with the scholarship office."
    },
    {
        "topic": "Course Change",
        "content": "Course changes require approval from the academic department. Students should contact the program office to confirm the applicable process and deadline."
    },
    {
        "topic": "Student Support",
        "content": "Students can contact the program office for academic or administrative support. They should describe the issue clearly and provide relevant non-sensitive details."
    }
]

print("Student policy knowledge base created!")
print("Total policy sections:", len(student_policies))
```

```python
# Step 7: Simple policy retrieval

def retrieve_policy(question):
    words = set(question.lower().split())
    results = []

    for policy in student_policies:
        text = (policy["topic"] + " " + policy["content"]).lower()
        score = sum(1 for word in words if len(word) > 3 and word in text)

        if score > 0:
            results.append((score, policy))

    results.sort(key=lambda item: item[0], reverse=True)

    if results:
        return [
            item[1] for item in results[:2]
        ]

    return ["No relevant policy found."]

# Test retrieval
print(retrieve_policy("What should I do if my fee payment fails?"))
```

```python
# Step 8: Persistent memory using a JSON file

import json
import re
from pathlib import Path

MEMORY_FILE = Path("/content/student_agent_memory.json")

def load_memory():
    if MEMORY_FILE.exists():
        return json.loads(MEMORY_FILE.read_text())
    return {}

def save_memory(key, value):
    # Never store passwords, OTPs, or similar secrets
    sensitive = re.compile(
        r"\b(password|passwd|otp|one.time.password|token|secret)\b",
        re.IGNORECASE
    )

    if sensitive.search(key) or sensitive.search(value):
        return "Sensitive information was not saved."

    memory = load_memory()
    memory[key] = value
    MEMORY_FILE.write_text(json.dumps(memory, indent=2))
    return "Memory saved successfully."

# Test useful memory
print(save_memory("programme", "PGDM"))
print(load_memory())
```

```python
# Step 9: Student Support Agent

def student_support_agent(question, student_id="STU001"):
    q = question.lower()

    # Decide which tool to use
    if "fee" in q or "payment" in q:
        result = check_fee_status(student_id)
        return f"Your fee information: {result}"

    elif "scholarship" in q:
        result = check_scholarship_status(student_id)
        return f"Your scholarship information: {result}"

    # Retrieve relevant policy information
    elif any(word in q for word in [
        "policy", "failed", "course", "deadline", "office"
    ]):
        result = retrieve_policy(question)
        return f"Relevant policy information: {result}"

    # Use remembered information
    elif "remember" in q or "programme" in q:
        memory = load_memory()
        return f"Your saved information: {memory}"

    else:
        return (
            "I couldn't identify the right tool for this question. "
            "Please ask about fees, scholarships, or student policies."
        )


# Test the agent
print(student_support_agent("What is my fee status?"))
print(student_support_agent("What is my scholarship status?"))
print(student_support_agent("What should I do if my fee payment failed?"))
```

```python
# Step 10: Tracing and evaluation

import time

agent_traces = []

def traced_agent(question, student_id="STU001"):
    start_time = time.time()

    answer = student_support_agent(question, student_id)

    latency = round(time.time() - start_time, 4)

    trace = {
        "question": question,
        "student_id": student_id,
        "answer": answer,
        "latency_seconds": latency,
        "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
    }

    agent_traces.append(trace)
    return answer

# Test questions
print(traced_agent("What is my fee status?"))
print(traced_agent("What is my scholarship status?"))

# Display traces
for trace in agent_traces:
    print(trace)
```

```python
# Step 11: Basic evaluation

test_cases = [
    ("What is my fee status?", "fee"),
    ("What is my scholarship status?", "scholarship"),
    ("What should I do if my fee payment failed?", "policy")
]

evaluation_results = []

for question, expected in test_cases:
    start = time.time()
    answer = student_support_agent(question)
    latency = round(time.time() - start, 4)

    passed = expected in answer.lower()

    evaluation_results.append({
        "question": question,
        "expected": expected,
        "passed": passed,
        "latency_seconds": latency
    })

for result in evaluation_results:
    print(result)

accuracy = sum(r["passed"] for r in evaluation_results) / len(evaluation_results) * 100
print(f"\nBasic task success rate: {accuracy:.1f}%")
```

```python
# STEP 12: Student Support Agent Chat

def chat():
    print("Student Support Agent")
    print("Type 'exit' to stop.\n")

    student_id = input("Enter student ID (STU001/STU002/STU003): ")
    student_id = student_id.strip().upper().rstrip(".")

    if student_id not in student_records:
        print("Invalid ID. Use STU001, STU002, or STU003.")
        return

    while True:
        question = input("\nYou: ").strip()

        if question.lower() == "exit":
            print("Agent: Goodbye!")
            break

        if not question:
            print("Please enter a question.")
            continue

        start = time.time()
        answer = student_support_agent(question, student_id)
        latency = round(time.time() - start, 4)

        print("Agent:", answer)

        agent_traces.append({
            "question": question,
            "student_id": student_id,
            "answer": answer,
            "latency_seconds": latency,
            "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
        })

chat()
```

```python
# Step 14: Define and run detailed tracing

import time
import json

detailed_traces = []

def detailed_agent(question, student_id="STU001"):
    start = time.time()
    q = question.lower()

    tool_used = "none"
    evidence = []

    if "fee" in q or "payment" in q:
        tool_used = "check_fee_status"
        answer = str(check_fee_status(student_id))

    elif "scholarship" in q:
        tool_used = "check_scholarship_status"
        answer = str(check_scholarship_status(student_id))

    elif any(word in q for word in ["policy", "failed", "course", "deadline", "office"]):
        tool_used = "retrieve_policy"
        evidence = retrieve_policy(question)
        answer = str(evidence)

    else:
        answer = student_support_agent(question, student_id)

    trace = {
        "question": question,
        "student_id": student_id,
        "tool_used": tool_used,
        "retrieved_evidence": evidence,
        "answer": answer,
        "latency_seconds": round(time.time() - start, 4)
    }

    detailed_traces.append(trace)
    return answer


print(detailed_agent("What is my fee status?", "STU001"))
print(detailed_agent("What should I do if my fee payment failed?", "STU001"))

with open("/content/detailed_traces.json", "w") as f:
    json.dump(detailed_traces, f, indent=2)

print("Step 14 complete!")
print("Detailed traces:", len(detailed_traces))
```

```python
# STEP 15: Fix policy retrieval

def retrieve_policy(question):
    q = question.lower()

    if any(word in q for word in ["fee", "payment", "paid", "portal"]):
        return [{
            "topic": "Fee Payment",
            "content": (
                "Check the official fee portal for dues. "
                "If payment fails, save the transaction reference "
                "and contact the accounts office. Confirm the next "
                "working day payment rule with the accounts office."
            )
        }]

    if "scholarship" in q:
        return [p for p in student_policies
                if p["topic"] == "Scholarship"]

    if "course" in q:
        return [p for p in student_policies
                if p["topic"] == "Course Change"]

    return ["No relevant policy found."]


print(retrieve_policy(
    "What should I do if my fee payment failed?"
))
```

```python
# STEP 16: Evaluate the three required tasks

import time

test_cases = [
    {
        "question": "What is my fee status?",
        "expected_tool": "fee"
    },
    {
        "question": "What is my scholarship status?",
        "expected_tool": "scholarship"
    },
    {
        "question": "What should I do if my fee payment failed?",
        "expected_tool": "policy"
    }
]

evaluation_results = []

for test in test_cases:
    question = test["question"]
    start = time.time()

    if test["expected_tool"] == "fee":
        answer = check_fee_status("STU001")
        passed = "fee_status" in answer

    elif test["expected_tool"] == "scholarship":
        answer = check_scholarship_status("STU001")
        passed = "scholarship_status" in answer

    else:
        answer = retrieve_policy(question)
        passed = (
            isinstance(answer, list)
            and len(answer) > 0
            and isinstance(answer[0], dict)
            and answer[0].get("topic") == "Fee Payment"
        )

    evaluation_results.append({
        "question": question,
        "expected": test["expected_tool"],
        "passed": passed,
        "latency_seconds": round(time.time() - start, 4)
    })

for result in evaluation_results:
    print(result)

accuracy = (
    sum(r["passed"] for r in evaluation_results)
    / len(evaluation_results) * 100
)

print(f"\nTask success rate: {accuracy:.1f}%")
```

```python
# STEP 22: Fix agent routing

def improved_agent(question, student_id="STU001"):
    start = time.time()
    q = question.lower()

    # Policy/action questions must be checked first
    if any(word in q for word in [
        "failed", "how do i", "what should i do",
        "policy", "deadline", "course change"
    ]):
        tool_used = "retrieve_policy"
        result = retrieve_policy(question)

    elif "scholarship" in q:
        tool_used = "check_scholarship_status"
        result = check_scholarship_status(student_id)

    elif any(word in q for word in ["fee", "payment", "dues"]):
        tool_used = "check_fee_status"
        result = check_fee_status(student_id)

    else:
        tool_used = "none"
        result = (
            "I cannot answer this from the available tools. "
            "Please contact the program office."
        )

    improved_traces.append({
        "question": question,
        "student_id": student_id,
        "tool_used": tool_used,
        "result": str(result),
        "latency_seconds": round(time.time() - start, 4)
    })

    return result


# Test the routing
improved_traces = []

print("Fee:", improved_agent("What is my fee status?"))
print("Scholarship:", improved_agent("What is my scholarship status?"))
print("Payment failure:", improved_agent(
    "What should I do if my fee payment failed?"
))

print("\nSelected tools:")
for trace in improved_traces:
    print(trace["question"], "->", trace["tool_used"])
```
